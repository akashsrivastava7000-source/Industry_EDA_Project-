# Univariate Analysis

#### Univariate Analysis means analyzing one variable at a time to understand its behavior and characteristics.
#### It helps to understand:

#### Distribution of data
#### Central tendency (mean, median, mode)
#### Spread of data
#### Outliers
#### Skewness
#### Kurtosis
#### standard Deviation

```python
# Continuous Variable Analysis
## select only numeric columns to find the distribution of data


num_cols = df.select_dtypes(include=['int64', 'float64']).columns

summary = pd.DataFrame({
    'Mean': df[num_cols].mean(),
    'Median': df[num_cols].median(),
    'Std Dev': df[num_cols].std(),
    'Min': df[num_cols].min(),
    'Max': df[num_cols].max(),
    'Skewness': df[num_cols].skew(),
    'Kurtosis': df[num_cols].kurt()
})

print(summary)
```

## Insights:
#### The statistical summary provides an overview of all numerical columns in the dataset.
#### Mean and median values help identify the central tendency of the data.
#### Standard deviation shows the spread or variability of each numerical feature.
#### Minimum and maximum values indicate the range of the data.
#### Skewness helps identify whether the data distribution is symmetric, positively skewed, or negatively skewed.
#### Kurtosis indicates the presence of outliers and the peakedness of the distribution.
#### Numerical columns with high standard deviation show greater variability in business performance.
#### Positive skewness in some variables suggests that a few companies have exceptionally high values compared to others.

```python
# using histogram to display Continuous Variable Analysis


num_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Different colors for each graph
colors = ['red', 'blue', 'green', 'orange', 'purple', 'brown', 'pink', 'cyan']

for i, col in enumerate(num_cols):
    
    # Calculate skewness and kurtosis
    skew_val = df[col].skew()
    kurt_val = df[col].kurt()
    
    plt.figure(figsize=(8,5))
    
    # Histogram with different colors
    sns.histplot(
        df[col],
        bins=30,
        kde=True,
        color=colors[i % len(colors)]
    )
    
    ## Show skewness and kurtosis on graph
    plt.text(
        0.95,
        0.95,
        f'Skewness = {skew_val:.2f}\nKurtosis = {kurt_val:.2f}',
        horizontalalignment='right',
        verticalalignment='top',
        transform=plt.gca().transAxes,
        bbox=dict(facecolor='white', alpha=0.7)
    )
    
    plt.title(f'Histogram of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    
    plt.show()
```

# Outlier Detection

Outliers are extreme values that differ significantly from the majority of observations.

In this project, outliers were detected using:
- Boxplots
- Interquartile Range (IQR) Method
- Z-score Method

Outlier analysis helps identify unusually large or small business values that may influence overall trends and statistical analysis.

## Boxplot Visualization
### Annual Revenue Boxplot

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x=df['annual_revenue_million'],
    color='gold'
)

plt.title('Boxplot of Annual Revenue',
          fontsize=15,
          color='darkred')

plt.xlabel('Annual Revenue (Million)',
           fontsize=12,
           color='darkblue')

plt.grid(color='lightgray',
         linestyle='--',
         alpha=0.7)

plt.show()
```

```python
# Employee Count Boxplot

plt.figure(figsize=(8,5))

sns.boxplot(
    x=df['employee_count'],
    color='lightgreen'
)

plt.title('Boxplot of Employee Count',
          fontsize=15,
          color='darkgreen')

plt.xlabel('Employee Count',
           fontsize=12,
           color='purple')

plt.grid(color='gray',
         linestyle=':',
         alpha=0.5)

plt.show()
```

```python
# Customer Count Boxplot

plt.figure(figsize=(8,5))

sns.boxplot(
    y=df['customer_count'],
    color='tomato'
)

plt.title('Boxplot of Customer Count',
          fontsize=15,
          color='maroon')

plt.xlabel('Customer Count',
           fontsize=12,
           color='navy')

plt.grid(color='silver',
         linestyle='-.',
         alpha=0.6)

plt.show()
```

```python
# IQR Method Code
# This method helps detect unusually high or low observations in the dataset.


# Select numerical columns
num_cols = df.select_dtypes(include=['int64','float64']).columns

# Calculate Quartiles
Q1 = df[num_cols].quantile(0.25)
Q3 = df[num_cols].quantile(0.75)

# Calculate Interquartile Range
IQR = Q3 - Q1

# Detect Outliers
outliers = ((df[num_cols] < (Q1 - 1.5 * IQR)) |
            (df[num_cols] > (Q3 + 1.5 * IQR)))

# Count Outliers
print(outliers.sum())
```

```python
# Z-Score Method
## This method helps identify extremely unusual observations in numerical variables.

from scipy.stats import zscore

## Calculate Z-score values
z_scores = np.abs(zscore(df[num_cols]))

## Detect outliers with threshold > 3
outlier_z = (z_scores > 3)

## Count Outliers
print(outlier_z.sum(axis=0))
```

# Observations

- Extreme values were present in annual_revenue_million.
- High outliers in employee_count represented multinational enterprises.
- Some companies showed exceptionally high customer_count, especially in the Retail sector.
- Outliers were retained because they represent genuine large-scale organizations and important business scenarios.

## Correlation matrix for numerical columns
#### A correlation matrix is a table that shows the correlation values between multiple variables in a dataset.

- How strongly they are related

#### implying that company age does not strongly impact revenue generation or market performance.

#### Overall, the analysis did not identify any severe multicollinearity among the numerical features,
